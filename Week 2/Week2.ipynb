```python
#Q1
import matplotlib.pyplot as plt
import numpy as np

fig= plt.figure()
ax=fig.add_axes([0,0,1,1])

x=np.arange(0, 100, 1)
y=np.arange(0, 200, 2)
ax.plot(x,y)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('title')

plt.show()
```

```python
#Q2
import matplotlib.pyplot as plt
import numpy as np

fig=plt.figure()

ax1=fig.add_axes([0,0,1,1])
ax2=fig.add_axes([0.2, 0.5, 0.2, 0.2])

x=np.arange(0, 100, 1)
y=np.arange(0, 200, 2)

ax1.plot(x,y)
ax2.plot(x,y)

plt.show()
```

```python
#Q3)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv('Week 2 company_sales_data.csv')

x=df['month_number']
y=df['total_profit']

plt.plot(x,y)
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.show()
```

```python
#Q4)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x=df['month_number']
y=df['total_profit']

plt.plot(x,y, linestyle='dotted', color='red', marker='o', markeredgecolor='red', linewidth=3, label='Profit data of last year')
plt.legend(loc='lower right')
plt.xlabel('Month Number')
plt.ylabel('Sold Units Number')
plt.show()
```

```python
# Additional Q1)
import matplotlib.pyplot as plt
import pandas as pd

x = df['month_number']
product_columns = [
    'facecream',
    'facewash',
    'toothpaste',
    'bathingsoap',
    'shampoo',
    'moisturizer'
]

plt.figure(figsize=(10, 6))

for product in product_columns:
  plt.plot(x, df[product], label=product)

plt.xlabel('Month Number')
plt.ylabel('Sales Units in Number')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

products = ['facecream', 'facewash', 'toothpaste',
            'bathingsoap', 'shampoo', 'moisturizer']

total_sales = df[products].sum()

plt.figure(figsize=(8,8))
plt.pie(total_sales,
        labels=products,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Total Sales of Each Product in Last Year")
plt.axis('equal')
plt.show()
```
